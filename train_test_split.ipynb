Step 1: Importing the tools we need

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

Step 2: Create/Load dataset

```python
data = {
    "Area": [1200, 1500, 1800, 1000, 2000, 1600, 1300, 2200, 1100, 1900],
    "Price": [50, 70, 85, 40, 110, 75, 55, 125, 42, 90]
}
```

```python
df=pd.DataFrame(data)
df.shape
```

Step 3: Separating the columns

```python
X=df[['Area']]
Y=df['Price']
```

Step 4: Keeping some data hidden for testing

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2)
```

Step 5: Training the model

```python
model = LinearRegression()
model.fit(X_train, Y_train)
print("\n Model Trained")
```

Step 6: asking price about new house

```python
new_house= pd.DataFrame({'Area':[600]})
price = model.predict(new_house)
print("price of 1600sqft house should cost:",round(price[0], 2),"lakhs")
```

```python
plt.scatter(X,Y,color="r")
plt.xlabel("Area in Sqft.")
plt.ylabel("Rate in Lakhs")
plt.grid()
plt.show
```
